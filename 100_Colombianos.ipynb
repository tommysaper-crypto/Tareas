Primero tenemos que instalar el faker y el paquete de polars tambien, utilizamos el signo "!" que permite ejecutar comandos del sistema en Colab.

```
import polars as pl
from faker import Faker
import matplotlib.pyplot as plt
import random
```

Tambien importamos los polars que nos siven para poder hacer el analisis y poder hacer manejo de las tablas que se generan con la información del faker, despues de hacer esto importamos la libreria de el faker para que nos ayude a generar los datos falsos para poder rellenar las variantes que tenemos en nuestra preguta sobre los 100 colombianos, la libreria matplotlib.pyplot, nos ayudara a hacer los gráficos que se muestren gracias a los datos que nos dara el faker, por último la libreria random que nos ayudara a generar números para las variantes como pueden ser las edades, los salarios entre otras.

```
url= "https://raw.githubusercontent.com/rfordatascience/tidytuesday/refs/heads/main/data/2023/2023-12-05/life_expectancy.csv"
df = pl.read_csv(url)
print(df.head())
print(df.columns)
```

En esta parte de el código podemos ver primero una variante la cual es url la cual contiene el link que dejo usted profe con los datos de los paises, para poder correr y manipular los datos para que queden los de los 100 colombianos, tambien se usan los polars para hacer las tablas y el csv que mira los valores separados por las comas, ademas se usa el df.head para mostrar los 5 primeros valores que existen en el link que se utilizo, y el d.columns para mostrar los nombres que tienen las columnas en los datos.

```
colombia = colombia = df.filter(
    pl.col("Entity") == "Colombia"
)
print(colombia.head())
```

Con el df[Entity] podemos ver que se eliminan las demas filas y deja solo las que hagan parte de Colombia como pide el ejercicio y para confirmar esto se puede utilizar el colombia.head para ver que los valores que se muestran aqui son los valores de Colombia.

```
print(colombia)
```

Tambien para continuar con el reconocimiento de las filas de Colombia se puede usar el colombia.info para saber mas información sobre las filas que tienen a Colombia en ellas como por ejemplo el tipo Dtype, que columna es y cual es su nombre entre más información que nos da la función.

```
print(colombia.null_count())
colombia = colombia.drop_nulls()
print(colombia.is_duplicated().sum())
colombia = colombia.unique()
```

Podemos seguir con la limpieza de nuestros datos con las siguientes funciones como por ejemplo el null_count, que detecta que vacios hay en los datos, tambien se puede encontrar en las otras funciones como quitar filas repetidas o duplicadas.

```
print(colombia.describe())
```

Podemos tambien usar la funcion colombia.describe para ver un resumen de las medidas estadisticas basicas que se pueden utilizar, esta función nos ahorra el tener que usar las funciones una por una, como por ejemplo el mean, median entre otras funciones que nos da automaticamente.

Para mejor entendimiento se puede decir que significa cada una de las variantes que nos da el describe empezando por count, que cuenta la cantidad de datos no vacios, mean que da el promedio de los valores, std que da la derivación estandar que es, que tan alejados estan los datos de los otros, min que es el valor minimo, 25%, 50%, 75%, que se refiere a los cuartiles aue hay en los datos y por ultimo max que es el valor maximo.

```
plt.plot(colombia["Year"], colombia["LifeExpectancy"])
plt.title("Esperanza de vida en Colombia")
plt.xlabel("Año")
plt.ylabel("Esperanza de vida")
plt.show()
```

Con el plot podemos hacer gráficas, las cuales son muy importantes para el mejor entendimiento de la estadistica que estamos generando, podemos ver las funciones que se necesitan para poder hacer una buena gráfica como por ejemplo, el xlabel y el ylabel que sirven para dar nombre a las variantes que hay en la tabla como por ejmplo el "año" o ""esperanza de vida", tambien es importante resaltar que este "plot" es lineal y se evidencia con lineas y puntos dentro de la gráfica.

```
plt.hist(colombia["LifeExpectancy"])
plt.title("Distribución de esperanza de vida")
plt.xlabel("Esperanza de vida")
plt.ylabel("Frecuencia")

plt.show()
```

Aqui podemos ver la misma situación ya que se usan las mismas funciones ya que se quiere mostrar, pero ahora con un histograma, se usa la misma información de la gráfica pasada.

```
fake = Faker('es_CO')
datos = []
```

En esta parte de el codigo se usa el faker y se ajusta a los terminos de Colombia con el ("es_CO"), y con esto podemos hacer que funcione de acuerdo a los datos colombianos, tambien se hace una lista vacia en la cual puedan entrar los datos colombianos para despues sacar las medidas de acuerdo a esta lista.

```
for i in range(100):
  persona = {
    "Nombre": fake.name(),
    "Edad": random.randint(18, 80),
    "Direccion": fake.address(),
    "Ciudad": fake.city(),
    "Telefono": fake.phone_number(),
    "Correo": fake.email(),
    "Salario": random.randint(1000000, 10000000),
    "Estado_Civil": random.choice([
      "Soltero",
      "Casado",
      "Divorciado"
    ])
  }
datos.append(persona)
```

Aqui usamos el "for i in range" para decir cuantas veces debe correr el mismo código, en este caso son 100 por los colombianos, despues creamos el diccionario "persona" para poder guardar los datos, ahi guardamos las varibles de nombre, edad, direccion, cuidad, telefono, correo, salario, y por ultimo estado civil, asi que este diccionario guarda muchas cosas importantes y nos ayuda con la información.

```
faker_df = pl.DataFrame(datos)
print(faker_df.head())
```

Hacemos un dataframe que es una función que nos da una lista de filas y columnas que se puede asimilar por ejemplo con una hoja de excel. Con los datos que nos dio el código anterior y usamos la terminación ".head" para mostrar los primeros 5 datos y que esto sea como una muestra de todos los datos presentes.

```
faker_df.write_csv("100_colombianos.csv")
print(faker_df.null_count())
faker_df = faker_df.unique()
faker_df = faker_df.with_columns(
    pl.col("Nombre").str.strip_chars()
)
```

Aca limpiamos los datos que nos dieron en los códigos anteriores, usando fuciones que nos dan los polars, asi que con estos quitamos los datos que estan duplicados o que se repiten, tambien, usamos print para poder ver, en la tabla la cantidad de elementos repetidos o desechados que hay, en este caso no se repite ninguno asi que la tabla da 0

```
print(faker_df.describe())
```

Con la información nueva podemos volver a sacar la función describe pero esta vez con el faker, este nos da la información de todas las columnas que hay, información sobre nombres, números, salarios, entre otras variantes, tambien medias, modas, minimos y maximos, tipo de dato y más información.

```
plt.hist(faker_df["Edad"])
plt.title("Distribución de edades")
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.show()
```

```
plt.hist(faker_df["Salario"])
plt.title("Distribución de salarios")
plt.xlabel("Salario")
plt.ylabel("Frecuencia")
plt.show()
```
