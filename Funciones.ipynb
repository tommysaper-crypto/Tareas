```
def greet():
    print("Hello, world!")
greet()
```

```
def show_name(name):
    print("Hello", name)
show_name("Tomas")
```

```
def add(a, b):
    return a + b
resultado = add(5, 3)
print(resultado)
```

```
def subtract(a, b):
    return a - b
print(subtract(10, 4))
```

```
def multiply(a, b):
    return a * b
print(multiply(3, 4))
```

```
def divide(a, b):
    return a / b
print(divide(10, 2))
```

```
def is_even(number):
    return number % 2 == 0
print(is_even(4))
```

```
def find_max(a, b):
    return max(a, b)
print(find_max(10, 20))
```

```
def square(number):
    return number ** 2
print(square(5))
```

```
def cube(number):
    return number ** 3
print(cube(3))
```

```
def count_letters(word):
    return len(word)
print(count_letters("python"))
```

```
def convert_to_upper(text):
    return text.upper()
print(convert_to_upper("hello"))
```

```
def calculate_area(length, width):
    return length * width
print(calculate_area(5, 3))
```

```
def average(numbers):
    return sum(numbers) / len(numbers)
print(average([10, 20, 30, 40]))
```

```
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
print(factorial(5))
```
